```python
import numpy as np
import matplotlib.pyplot as plt




```

```python

def vplotter(k, theta):
    x1 = np.cos(theta)
    x2 = np.sin(theta)
    return 0.5 * (1 - np.cos((k/2) * (1 - x1)) * x1 + np.sin((k/2) * (1 - x1)) * x2)
```

```python
theta = np.linspace(0, 2 * np.pi, 500)
k1 = 1.5
k2 = -1.5

plt.plot(theta, vplotter(k1, theta), label=f'V1')
plt.plot(theta, vplotter(k2, theta), label=f'V2')

plt.xlabel("Theta(rads)")
plt.ylabel("V(T)")
plt.title("V over theta")
plt.legend()
plt.grid()
plt.show()
```
